```python
import json
import os

scene_path = "/content/ycbv/test/test/000057"

files = {
    "Ground truth": "scene_gt.json",
    "Visibility information": "scene_gt_info.json",
    "Camera information": "scene_camera.json"
}

data = {}

for name, filename in files.items():
    path = os.path.join(scene_path, filename)

    with open(path, "r") as f:
        data[name] = json.load(f)

    print(f"\n{name}")
    print("Available image IDs:", len(data[name]))
    print("First image ID:", list(data[name].keys())[0])
    print("First entry:", data[name][list(data[name].keys())[0]])
```

```python
# Check that each object appears in every image of a scene

from collections import Counter

gt = data["Ground truth"]

object_counts = Counter()

for image_id, annotations in gt.items():
    for annotation in annotations:
        object_counts[annotation["obj_id"]] += 1

print("Object appearances in scene 000057:\n")

for obj_id, count in sorted(object_counts.items()):
    print(f"Object ID {obj_id}: {count} appearances")
```

```python
import numpy as np
from collections import Counter

gt = data["Ground truth"]
camera = data["Camera information"]

# Check whether each object category occurs only once in each image.
duplicates = []

for image_id, annotations in gt.items():
    counts = Counter(a["obj_id"] for a in annotations)

    for obj_id, count in counts.items():
        if count > 1:
            duplicates.append((image_id, obj_id, count))

print("Images inspected:", len(gt))
print("Images with duplicate object categories:", len(duplicates))

# Check camera movement between the first and last images.
image_ids = sorted(gt.keys(), key=int)

first = camera[image_ids[0]]
last = camera[image_ids[-1]]

R1 = np.array(first["cam_R_w2c"]).reshape(3, 3)
R2 = np.array(last["cam_R_w2c"]).reshape(3, 3)

t1 = np.array(first["cam_t_w2c"])
t2 = np.array(last["cam_t_w2c"])

# Camera position in world coordinates
C1 = -R1.T @ t1
C2 = -R2.T @ t2

print("\nCamera position in first image:", C1)
print("Camera position in last image:", C2)
print("Camera translation difference:",
      np.linalg.norm(C2 - C1), "mm")

# Camera orientation difference
R_relative = R2 @ R1.T

cos_angle = np.clip(
    (np.trace(R_relative) - 1) / 2,
    -1, 1
)

angle = np.degrees(np.arccos(cos_angle))

print("Camera orientation difference:", angle, "degrees")
```

```python
from collections import defaultdict

scene_path = "/content/ycbv/test/test/000057"

def load_json(filename):
    with open(os.path.join(scene_path, filename), "r") as f:
        return json.load(f)

gt = load_json("scene_gt.json")
camera = load_json("scene_camera.json")

# Store world-coordinate poses for each object category
world_poses = defaultdict(list)

for image_id, annotations in gt.items():

    cam_data = camera[image_id]

    R_w2c = np.array(
        cam_data["cam_R_w2c"]
    ).reshape(3, 3)

    t_w2c = np.array(
        cam_data["cam_t_w2c"]
    ).reshape(3, 1)

    # Transform from camera coordinates to world coordinates
    T_w2c = np.eye(4)
    T_w2c[:3, :3] = R_w2c
    T_w2c[:3, 3] = t_w2c.flatten()

    T_c2w = np.linalg.inv(T_w2c)

    for annotation in annotations:

        obj_id = annotation["obj_id"]

        R_m2c = np.array(
            annotation["cam_R_m2c"]
        ).reshape(3, 3)

        t_m2c = np.array(
            annotation["cam_t_m2c"]
        ).reshape(3, 1)

        T_c2o = np.eye(4)
        T_c2o[:3, :3] = R_m2c
        T_c2o[:3, 3] = t_m2c.flatten()

        # Object pose in world coordinates
        T_w2o = T_c2w @ T_c2o

        world_poses[obj_id].append(
            (int(image_id), T_w2o)
        )


def rotation_difference(R1, R2):
    relative_R = R1.T @ R2

    cos_angle = (
        np.trace(relative_R) - 1
    ) / 2

    cos_angle = np.clip(cos_angle, -1, 1)

    return np.degrees(np.arccos(cos_angle))


print("Object Pose Consistency\n")

for obj_id, observations in sorted(world_poses.items()):

    positions = np.array([
        T[:3, 3] for _, T in observations
    ])

    reference_R = observations[0][1][:3, :3]

    position_center = np.median(positions, axis=0)

    position_deviations = np.linalg.norm(
        positions - position_center,
        axis=1
    )

    rotation_deviations = [
        rotation_difference(reference_R, T[:3, :3])
        for _, T in observations
    ]

    print(f"Object ID: {obj_id}")
    print(f"Observations: {len(observations)}")
    print(
        f"Maximum position deviation: "
        f"{position_deviations.max():.4f} mm"
    )
    print(
        f"Maximum rotation deviation: "
        f"{max(rotation_deviations):.4f} degrees"
    )
    print()
```

```python
# Extract crops using bbox

from PIL import ImageDraw

rgb_path = os.path.join(scene_path, "rgb")

# Candidate pairs: object ID -> (occluded image, clearer image)
pairs = {
    20: (1827, 1440),
    21: (1827, 1108)
}

def extract_object_crop(image_id, obj_id, margin=0.35):

    # Load original RGB image
    image_path = os.path.join(
        rgb_path, f"{image_id:06d}.png"
    )
    image = Image.open(image_path).convert("RGB")

    # Find the target object's annotation
    annotations = gt[str(image_id)]
    image_info = info[str(image_id)]

    for annotation, annotation_info in zip(
        annotations, image_info
    ):
        if annotation["obj_id"] == obj_id:

            # Full projected object bounding box
            x, y, w, h = annotation_info["bbox_obj"]

            break
    else:
        raise ValueError(
            f"Object {obj_id} not found in image {image_id}"
        )

    # Expand the bounding box
    margin_x = w * margin
    margin_y = h * margin

    left = max(0, int(np.floor(x - margin_x)))
    top = max(0, int(np.floor(y - margin_y)))

    right = min(
        image.width,
        int(np.ceil(x + w + margin_x))
    )

    bottom = min(
        image.height,
        int(np.ceil(y + h + margin_y))
    )

    # Extract the crop from the ORIGINAL image
    crop = image.crop((left, top, right, bottom))

    # Draw boxes for visual inspection
    annotated_image = image.copy()
    draw = ImageDraw.Draw(annotated_image)

    # Original object bounding box
    draw.rectangle(
        (int(x), int(y), int(x + w), int(y + h)),
        outline="red",
        width=3
    )

    # Expanded crop region
    draw.rectangle(
        (left, top, right, bottom),
        outline="lime",
        width=3
    )

    return annotated_image, crop


# Display original images and their extracted crops
fig, axes = plt.subplots(4, 2, figsize=(12, 20))

for row, (obj_id, image_ids) in enumerate(pairs.items()):

    for image_id in image_ids:

        col = 0 if image_id == image_ids[0] else 1

        annotated, crop = extract_object_crop(
            image_id, obj_id, margin=0.35
        )

        axes[row * 2 + col, 0].imshow(annotated)
        axes[row * 2 + col, 0].set_title(
            f"Object {obj_id} | Image {image_id}\n"
            "Red: object box | Green: expanded crop"
        )
        axes[row * 2 + col, 0].axis("off")

        axes[row * 2 + col, 1].imshow(crop)
        axes[row * 2 + col, 1].set_title(
            f"Extracted crop | Object {obj_id}"
        )
        axes[row * 2 + col, 1].axis("off")

plt.tight_layout()
plt.show()
```

```python
# DINOv2 input settings
IMAGE_SIZE = 518

# ImageNet channel statistics used for normalization
IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD = (0.229, 0.224, 0.225)


def preprocess_dinov2(image, image_size=IMAGE_SIZE):

    # Ensure the image has three RGB channels
    image = image.convert("RGB")

    # Get the original width and height
    width, height = image.size

    # Find the size of the square canvas
    square_size = max(width, height)

    # Create a square canvas.
    # Its colour is set to the ImageNet mean so that the padding becomes approximately zero after normalization.
    mean_colour = tuple(
        round(value * 255) for value in IMAGENET_MEAN
    )

    square_image = Image.new(
        "RGB",
        (square_size, square_size),
        mean_colour
    )

    # Calculate where to place the original image
    left = (square_size - width) // 2
    top = (square_size - height) // 2

    # Place the original image in the centre so no stretching or distortion occurs here.
    square_image.paste(image, (left, top))

    # Resize the square image to 518 x 518
    resized_image = square_image.resize(
        (image_size, image_size),
        Image.Resampling.BICUBIC
    )

    # Convert image pixels into a NumPy array
    image_array = np.array(resized_image).astype(np.float32)

    # Scale pixel values from [0, 255] to [0, 1]
    image_array = image_array / 255.0

    # Convert from H x W x C to C x H x W
    image_tensor = torch.from_numpy(
        image_array
    ).permute(2, 0, 1)

    # Normalize each RGB channel
    mean = torch.tensor(IMAGENET_MEAN).view(3, 1, 1)
    std = torch.tensor(IMAGENET_STD).view(3, 1, 1)

    image_tensor = (image_tensor - mean) / std

    return resized_image, image_tensor
```

```python
# Apply preprocessing to the four candidate crops

# Candidate image pairs
candidate_images = [
    (20, 1827),
    (20, 1440),
    (21, 1827),
    (21, 1108)
]

processed_images = {}

for obj_id, image_id in candidate_images:

    # Extract the original RGB crop using existing function
    annotated_image, crop = extract_object_crop(
        image_id,
        obj_id,
        margin=0.35
    )

    # Apply DINOv2 preprocessing
    resized_image, image_tensor = preprocess_dinov2(crop)

    # Save for later use
    processed_images[(obj_id, image_id)] = {
        "crop": crop,
        "resized": resized_image,
        "tensor": image_tensor
    }

    print(
        f"{object_names[obj_id]} | Image {image_id}"
    )
    print("Original crop:", crop.size)
    print("Preprocessed tensor:", image_tensor.shape)
    print()
```

```python
# Download the Dinov2 model

model = torch.hub.load(
    'facebookresearch/dinov2',
    'dinov2_vits14'
)

model.eval()
```

```python
# Test Dinov2

# Select GPU if available, otherwise use CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Move the model to the selected device
model = model.to(device)
model.eval()

# Stack the four image tensors into one batch
batch = torch.stack([
    data["tensor"]
    for data in processed_images.values()
]).to(device)

print("Batch shape:", batch.shape)
print("Device:", device)

# Extract DINOv2 features
with torch.inference_mode():
    features = model.forward_features(batch)["x_norm_clstoken"]

print("\nFeature shape:", features.shape)
print("Feature data type:", features.dtype)
print("Contains NaN:", torch.isnan(features).any().item())
print("Contains infinity:", torch.isinf(features).any().item())
```

```python
# Examine how similar the extracted features are

import torch.nn.functional as F

# Preserve the order of the four images
feature_labels = [
    f"{object_names[obj_id]}\nImage {image_id}"
    for obj_id, image_id in processed_images.keys()
]

# Normalize each feature vector
normalized_features = F.normalize(features, p=2, dim=1)

# Calculate pairwise cosine similarity
similarity_matrix = normalized_features @ normalized_features.T

# Convert to NumPy for plotting
similarity_matrix = similarity_matrix.cpu().numpy()

# Display results
fig, ax = plt.subplots(figsize=(9, 7))

image = ax.imshow(similarity_matrix, cmap="viridis", vmin=-1, vmax=1)

ax.set_xticks(range(len(feature_labels)))
ax.set_yticks(range(len(feature_labels)))
ax.set_xticklabels(feature_labels, rotation=35, ha="right")
ax.set_yticklabels(feature_labels)

for i in range(len(feature_labels)):
    for j in range(len(feature_labels)):
        ax.text(
            j, i,
            f"{similarity_matrix[i, j]:.3f}",
            ha="center",
            va="center",
            color="white"
        )

plt.colorbar(image, ax=ax, label="Cosine similarity")
plt.title("Pairwise DINOv2 Feature Similarity")
plt.tight_layout()
plt.show()
```

```python
# Print the four important comparisons

comparisons = [
    (0, 1, "Extra Large Clamp: occluded vs visible"),
    (2, 3, "Foam Brick: occluded vs visible"),
    (0, 2, "Different objects: both occluded"),
    (1, 3, "Different objects: both more visible")
]

for i, j, description in comparisons:
    similarity = similarity_matrix[i, j]

    print(f"{description}: {similarity:.4f}")
```
